```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python


n = 500  # number of customers

data = {
    "Customer_ID": range(1, n+1),
    "Age": np.random.randint(18, 70, n),
    "Income": np.random.randint(20000, 100000, n),
    "Kidhome": np.random.randint(0, 3, n),
    "Teenhome": np.random.randint(0, 2, n),
    "Recency": np.random.randint(1, 100, n),

    "MntWines": np.random.randint(0, 1000, n),
    "MntFruits": np.random.randint(0, 500, n),
    "MntMeat": np.random.randint(0, 1000, n),

    "NumWebPurchases": np.random.randint(0, 20, n),
    "NumStorePurchases": np.random.randint(0, 20, n),

    "Response": np.random.randint(0, 2, n)
}

df = pd.DataFrame(data)
df.to_csv("customer_data.csv", index=False)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
from sklearn.model_selection import train_test_split
X=df.drop("Disease",axis=1)
Y=df.Disease
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.15,random_state=42)
```

```python
x_train.shape,y_train.shape
```

```python
df.Disease.value_counts(),y_train.value_counts()
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)

pred = lr.predict(x_test)

from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score

accuracy_score(pred,y_test) * 100
```

```python

w = lr.coef_[0]
b = lr.intercept_[0]

x_range = np.linspace(df.Cholesterol.min(), df.Cholesterol.max(), 30)
y_range = np.linspace(df.Resting_BP.min(), df.Resting_BP.max(), 30)

xx, yy = np.meshgrid(x_range, y_range)


zz = -(w[0]*xx + w[1]*yy + b) / w[2]

# Plot points
fig2 = go.Figure()

fig2.add_trace(go.Scatter3d(
    x=df.Cholesterol,
    y=df.Resting_BP,
    z=df.Age,
    mode='markers',
    marker=dict(
        size=4,
        color=df.Disease,
        
        opacity=0.8
    ),
    name="Data"
))

# Add hyperplane
fig2.add_trace(go.Surface(
    x=xx,
    y=yy,
    z=zz,
    opacity=0.5,
    name="Decision Plane"
))

fig2.update_layout(
    title="3D Logistic Regression Hyperplane",
    scene=dict(
        xaxis_title='Cholesterol',
        yaxis_title='Resting BP',
        zaxis_title='Age'
    ),
    margin=dict(l=0, r=0, t=20, b=0)
)

fig2.show()
```

```python
fig2.write_html("scatter with hyper plane.html")
fig2.write_image("Hyper plane.jpeg")
```

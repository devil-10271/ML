# One hot Encoding using PANDAS

```python
pd.get_dummies(df,columns=['owner','fuel'],dtype=np.int32)
```

```python
pd.get_dummies(df,columns=['owner','fuel'],dtype=np.int32,drop_first='first')
```

# One hot Encoding using SKLEARN.PREPROCESSING

```python
X=df.iloc[:,0:4]
Y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(drop="first",sparse_output=True)
```

```python
ohe.fit_transform(x_train[['fuel','owner']])
```

```python
np.array(df[['brand','km_driven']])
```

```python
pd.DataFrame(np.hstack((np.array(x_train[['brand','km_driven']]),ohe.transform(x_train[['fuel','owner']]).toarray())))
```

```python
pd.DataFrame(np.hstack((np.array(x_test[['brand','km_driven']]),ohe.transform(x_test[['fuel','owner']]).toarray())))
```

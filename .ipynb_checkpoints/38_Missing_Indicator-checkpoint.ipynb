```python
import numpy as np 
import pandas as pd 

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
```

```python
df= pd.read_csv('./CSVs/Titanic-Dataset.csv')
```

```python
df
```

```python
df.drop(['PassengerId','Name','Cabin','Ticket'],axis=1,inplace=True)
```

```python
df
```

```python
df.isnull().mean()*100
```

```python
X=df.drop('Survived',axis=1)
```

```python
Y=df.Survived
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
num_col=['Age','Fare']

num_col_pip=Pipeline(
    steps=[
        ('Imputer',SimpleImputer(strategy='median')),
        ('Scaling',StandardScaler())
    ]
)
```

```python
cat_col=['Sex','Embarked']

cat_col_pip=Pipeline(
    steps=[
        ('Imputer_ca',SimpleImputer(strategy='most_frequent')),
        ('OHE',OneHotEncoder(handle_unknown='ignore'))
    ]
)
```

```python
preprocess=ColumnTransformer(
    transformers=[
        ('num',num_col_pip,num_col),
        ('cat',cat_col_pip,cat_col)
    ]      
)
```

```python
clf=Pipeline(steps=[
    ('proc',preprocess),
    ('model',LogisticRegression())
])
```

```python
from sklearn import set_config
```

```python
set_config(display='diagram')
clf
```

```python
param_grid = { 
    # Use 'proc' instead of 'preprocessor'
    'proc__num__Imputer__strategy': ['mean', 'median'], 
    'proc__cat__Imputer_ca__strategy': ['most_frequent', 'constant'], 
    
    # Use 'model' instead of 'classifier'
    'model__C': [0.1, 1.0, 10, 100] 
}
grid_search=GridSearchCV(clf,param_grid,cv=10)
```

```python
grid_search.fit(x_train,y_train)
```

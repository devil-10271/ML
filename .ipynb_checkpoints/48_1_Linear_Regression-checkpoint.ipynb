```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
np.random.seed(32)  # For reproducibility
X = 2 * np.random.rand(1000, 1)  # 100 random points in the range [0, 2]
y = 4 + 3 * X + np.random.randn(1000, 1)  # Linear relation with noise

# Combine into a DataFrame for easier plotting
df = pd.DataFrame(np.hstack((X, y)), columns=['X', 'Y'])
```

```python
sns.scatterplot(x=df.X,y=df.Y)
```

```python
X,Y= df.X,df.Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,random_state=10,test_size=0.15)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train.values.reshape(-1,1),y_train)
```

```python
lr.predict(x_test.iloc[1].reshape(-1,1))
```

```python
sns.scatterplot(x=X,y=Y)
plt.plot(x_train,lr.predict(x_train.values.reshape(-1,1)),color="red")
plt.xlabel("X - axis ")
plt.ylabel("Y - axis ")
plt.show()
```

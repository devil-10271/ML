```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge,LinearRegression
from sklearn.metrics import r2_score
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_features=1 ,n_samples=500,n_targets=1,noise=18,random_state=13) 
```

```python
plt.scatter(x,y)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=10,test_size=0.15)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
rid_10= Ridge(alpha=10)
rid_10.fit(x_train,y_train)
```

```python
rid_100= Ridge(alpha=100)
rid_100.fit(x_train,y_train)
```

```python
pred1 = lr.predict(x_test)
pred2 = rid_10.predict(x_test)
pred3 = rid_100.predict(x_test)
```

```python
print(r2_score(y_test,pred1))
print(r2_score(y_test,pred2))
print(r2_score(y_test,pred3))
```

```python
plt.plot(x,y,'b.')
plt.plot(x,lr.predict(x),color="red",label="alpha 0")
plt.plot(x,rid_10.predict(x),color="orange",label="alpha 10")
plt.plot(x,rid_100.predict(x),color="yellow",label="alpha 100")
plt.legend()
plt.show()
```

```python
class meraLR:
    def __init__(self,alpha):
        self.coff_ = None
        self.intercept_ = None
        self.alpha = alpha
    
    def fit(self,x_train,y_train):
        x_train = np.array(x_train)
        y_train = np.array(y_train)
        
        x_mean = x_train.mean()
        y_mean = y_train.mean()
        
        # Vectorized calculation (No loops needed!)
        num = np.sum((y_train - y_mean) * (x_train - x_mean))
        den = np.sum((x_train - x_mean) * (x_train - x_mean))
        
        # Apply the Ridge regularization penalty (alpha) to the denominator
        self.coff_ = num / (den + self.alpha)
        self.intercept_ = y_mean - (self.coff_ * x_mean)
        
        print(f"Coefficient: {self.coff_} , Intercept: {self.intercept_}")
        
    
    def predict(self,x_test):
        return self.intercept_  +  (self.coff_ * x_test)
```

```python
RID = meraLR(alpha=10)
```

```python
RID.fit(x_train,y_train),rid_10.coef_ ,rid_10.intercept_
```

```python
pred4 = RID.predict(x_test)
```

```python
r2_score(y_test,pred4)
```

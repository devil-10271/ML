```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,ElasticNet,Lasso,Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x,y = load_diabetes(return_X_y=True)
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.15)
```

```python
Lin_Reg = LinearRegression()
Lin_Reg.fit(x_train,y_train)
lin_reg_pred = Lin_Reg.predict(x_test)
r2_score(y_test,lin_reg_pred)
```

```python
Rid_Reg = Ridge(alpha=0.3)
Rid_Reg.fit(x_train,y_train)
Rid_reg_pred = Rid_Reg.predict(x_test)
r2_score(y_test,Rid_reg_pred)
```

```python
Lass_Reg = Lasso(alpha=0.1)
Lass_Reg.fit(x_train,y_train)
Lass_reg_pred = Lass_Reg.predict(x_test)
r2_score(y_test,Lass_reg_pred)
```

```python
Elas_Reg = ElasticNet(alpha=0.001,l1_ratio=0.5)
Elas_Reg.fit(x_train,y_train)
Elas_reg_pred = Elas_Reg.predict(x_test)
r2_score(y_test,Elas_reg_pred)
```

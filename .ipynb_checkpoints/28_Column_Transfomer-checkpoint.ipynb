```python
from sklearn.compose import ColumnTransformer
```

```python
import pandas as pd 
import numpy as np 
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
```

```python
import pandas as pd
```

```python
df=pd.read_csv("./CSVs/covid_toy.csv")
```

# by using old method 

```python
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
age=df.drop(['gender','fever','cough','city','has_covid'],axis=1).values
```

```python
ohe=OneHotEncoder(drop="first",sparse_output=False)
gender=ohe.fit_transform(df[['gender']])
```

```python
si=SimpleImputer()
fever=si.fit_transform(df[['fever']])
```

```python
od=OrdinalEncoder(categories=[["Mild","Strong"]])
```

```python
cough=od.fit_transform(df[['cough']])
```

```python
city=ohe.fit_transform(df[['city']])
```

```python
X=pd.DataFrame(np.concatenate((age,gender,fever,cough,city),axis=1))
```

```python
X
```

# From Column Transfomer

```python
ct=ColumnTransformer(transformers=[
    ('for_gender',OneHotEncoder(drop="first",sparse_output=False),['gender']),
    ('for_fever',SimpleImputer(),['fever']),
    ('for_cough',OrdinalEncoder(categories=[['Mild',"Strong"]]),['cough']),
    ('for_city',OneHotEncoder(drop="first",sparse_output=False),['city']),
],remainder="passthrough")
```

```python
X_ct=ct.fit_transform(df)
```

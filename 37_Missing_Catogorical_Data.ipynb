```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("./CSVs/train.csv",usecols=['FireplaceQu','GarageQual','SalePrice'])
```

```python
df.info()
```

```python
df.isnull().mean()*100
```

```python
df.FireplaceQu.value_counts().plot(kind='bar')
```

```python
df.GarageQual.value_counts().plot(kind='bar')
```

```python
fig=plt.figure()
sx=fig.add_subplot(111)

df[df.GarageQual == 'TA']['SalePrice'].plot(kind="kde", ax=sx)
df[df.GarageQual.isnull()]['SalePrice'].plot(kind="kde", ax=sx)

lines,labels=sx.get_legend_handles_labels()
labels=["House with TA","House With Null value"]
sx.legend(lines,labels,loc='best')
```

```python
temp=df[df.GarageQual == 'TA' ]['SalePrice']
```

```python
df.GarageQual.fillna('TA',inplace=True)
```

```python
df.GarageQual.value_counts().plot(kind='bar')
```

```python
fig=plt.figure()
sax=fig.add_subplot(111)

temp.plot(kind='kde',ax=sax)
df[df.GarageQual == 'TA']['SalePrice'].plot(kind="kde", ax=sax)

lines,labels=sax.get_legend_handles_labels()
labels=["original","replace value"]
sax.legend(lines,labels,loc='best')
```

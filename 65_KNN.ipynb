```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("./CSVs/Breast Cancer.csv")
```

```python
df
```

```python
df.drop(["id","Unnamed: 32"],axis=True,inplace=True)
```

```python
df
```

```python
le = LabelEncoder()
```

```python
import plotly.express as px
```

```python
df.diagnosis.value_counts()
```

```python
fig=px.pie(names=["0 no cancer","1 Have Cancer"],values=df.diagnosis.value_counts(),title="Percentage of Cancer patient",height=400,width = 400)
fig.show()
```

```python
df.diagnosis = le.fit_transform(df.diagnosis)
```

```python
df.shape
```

```python
SS = StandardScaler()
```

```python
x = df.drop("diagnosis",axis=1)
y= df.diagnosis
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.15)
```

```python
x_train = SS.fit_transform(x_train)
```

```python
x_test = SS.transform(x_test)
```

```python
x_train[0,:]
```

```python
knn = KNeighborsClassifier(n_neighbors=4)
knn.fit(x_train,y_train)
```

```python
pred = knn.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,pred)
```

```python
scores = list()
for i in range(1,20):
    KNN = KNeighborsClassifier(n_neighbors=i)
    KNN.fit(x_train,y_train)
    pred = KNN.predict(x_test)
    scores.append(accuracy_score(y_test,pred))
```

```python
fig = px.line(scores)
fig.show()
```

```python
# k is best at value 2,4,5,6,7,8
```

```python
from sklearn.metrics import f1_score
```

```python
scores = list()
for i in range(1,20):
    KNN = KNeighborsClassifier(n_neighbors=i)
    KNN.fit(x_train,y_train)
    pred = KNN.predict(x_test)
    scores.append(f1_score(y_test,pred))
```

```python
fig = px.line(scores)
fig.show()
```

```python
# from f1 score it is 12 is best value of k 
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x,y)
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(x.values,y.astype("int").values,clf=lr,legend=2)
plt.show()
```

```python
from sklearn.model_selection import cross_val_score

cross_val_score(lr,x,y,scoring="accuracy",cv=10).mean()*100
```

```python
clf1 = LogisticRegression()
np.mean(cross_val_score(clf1,X_trf,y,scoring='accuracy',cv=10))
```

```python
def plot_decision_boundary(X, y, degree=1):
    # Convert DataFrame/Series input to NumPy array for 2D slicing
    X = np.asarray(X)
    y = np.asarray(y)

    poly = PolynomialFeatures(degree=degree)
    X_trf = poly.fit_transform(X)

    clf = LogisticRegression()
    clf.fit(X_trf, y)

    accuracy = np.mean(
        cross_val_score(clf, X_trf, y, scoring='accuracy', cv=10)
    )

    a = np.arange(start=X[:, 0].min() - 1, stop=X[:, 0].max() + 1, step=0.01)
    b = np.arange(start=X[:, 1].min() - 1, stop=X[:, 1].max() + 1, step=0.01)

    XX, YY = np.meshgrid(a, b)

    # Simplified 2D grid array creation
    input_array = np.c_[XX.ravel(), YY.ravel()]

    labels = clf.predict(poly.transform(input_array))

    plt.contourf(XX, YY, labels.reshape(XX.shape), alpha=0.5)
    plt.scatter(X[:, 0], X[:, 1], c=y)
    plt.title('Degree = {}, accuracy is {}'.format(degree, np.round(accuracy, 4)))
```

```python
for i in range(1,8):
    plt.figure(figsize=(5,3))
    plot_decision_boundary(x,y,degree=i)
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as ex
import seaborn as sns
import plotly.express as px

```

```python
df = pd.read_csv("diabetes.csv")
```

```python
df_cols= df.columns
df.describe()
```

```python
features = [col for col in df.columns if col != 'Outcome']

# Create and show a scatter plot for each consecutive pair of features
for i in range(len(features) - 1):
    x_col = features[i]
    y_col = features[i + 1]
    
    fig = px.scatter(
        df,
        x=x_col,
        y=y_col,
        color='Outcome',
        title=f'{x_col} vs {y_col} (colored by Outcome)',
        labels={'Outcome': 'Diabetes Outcome'},
        color_continuous_scale='Viridis',  # or use 'Rainbow' for discrete? Outcome is numeric 0/1
        hover_data=df.columns.tolist()     # show all values on hover
    )
    # Improve discrete color mapping (0 and 1)
    fig.update_traces(marker=dict(size=8, opacity=0.7))
    fig.update_layout(
        xaxis_title=x_col,
        yaxis_title=y_col,
        legend_title_text='Outcome',
        template='plotly_white'
    )
    fig.show()
```

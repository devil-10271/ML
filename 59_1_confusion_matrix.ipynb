```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("./CSVs/heart_disease_dataset.csv")
```

```python
df.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
model1 = LogisticRegression()
model2 = DecisionTreeClassifier()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop("Disease",axis=1)
y=df.Disease
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
model1.fit(x_train,y_train)
model2.fit(x_train,y_train)
```

```python
pred_1 = model1.predict(x_test)
pred_2 = model2.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,precision_score,recall_score,accuracy_score,f1_score
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))


sns.heatmap(confusion_matrix(y_test,pred_1),annot=True,ax=ax1)
ax1.set_title('Logistic Regression')
ax1.set_xlabel('X axis')
ax1.set_ylabel('Y axis')


sns.heatmap(confusion_matrix(y_test,pred_2),annot=True,ax=ax2)
ax2.set_title('Decision Tree')
ax2.set_xlabel('X axis')
ax2.set_ylabel('Y axis')


plt.tight_layout()
plt.show()
```

```python
print(f"Logistic Regression accuracy : {accuracy_score(y_test,pred_1)}")
print(f"Logistic Regression precision : {precision_score(y_test,pred_1)}")
print(f"Logistic Regression recall : {recall_score(y_test,pred_1)}")
print(f"Logistic Regression f1 score : {f1_score(y_test,pred_1)}")

```

```python
print(f"Decision tree accuracy : {accuracy_score(y_test,pred_2)}")
print(f"Decision tree precision : {precision_score(y_test,pred_2)}")
print(f"Decision tree recall : {recall_score(y_test,pred_2)}")
print(f"Decision tree f1 score : {f1_score(y_test,pred_2)}")

```

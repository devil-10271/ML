```python
class apna_lr:
    def __init__(self):
        self.m = None
        self.b = None
        
    def fit(self,x_tr,y_tr):
        x_tr = x_tr.values
        y_tr = y_tr.values
        num = 0
        den = 0
        for i in range(x_tr.shape[0]):
            num += (x_tr[i] - x_tr.mean()) * (y_tr[i] - y_tr.mean())
            den += (x_tr[i] - x_tr.mean()) * (x_tr[i] - x_tr.mean())
        
        # print(num,den)
        self.m = num / den
        self.b = y_tr.mean() - (self.m * x_tr.mean())
        
        print(f' Slope ( m ) : {self.m}\n Intercept ( b ) : {self.b}')

        
    def predict(self,x_te):
        x_te = x_te.values
        return (self.m * x_te) + self.b
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(32)  # For reproducibility
X = 2 * np.random.rand(1000, 1)  # 100 random points in the range [0, 2]
y = 4 + 3 * X + np.random.randn(1000, 1)  # Linear relation with noise

# Combine into a DataFrame for easier plotting
df = pd.DataFrame(np.hstack((X, y)), columns=['X', 'Y'])
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.X,df.Y,test_size=0.15,random_state=42)
```

```python
lr = apna_lr()
```

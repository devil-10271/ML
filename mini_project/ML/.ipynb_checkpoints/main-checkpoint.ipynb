```python
x=df.drop(columns="Heart_ stroke")
```

```python
y=df["Heart_ stroke"]
```

```python
df.shape , x.shape, y.shape
```

```python
trf1 = ColumnTransformer([
    ("SI_1",SimpleImputer(strategy="most_frequent"),["education"]),
    ("SI_2",SimpleImputer(),[4,5,9,12,13,14])],
    remainder="passthrough"
)
```

```python
## Encoder
trf2= ColumnTransformer(
    [
        ("OHE_Gender",OneHotEncoder(sparse_output=False,drop="first"),[7]),
        ("OE_Education",OrdinalEncoder(categories=[["uneducated","primaryschool","graduate","postgraduate"]]),[0]),
        ("OE_PrevStorke",OrdinalEncoder(categories=[["no","yes"]]),[10])
    ],remainder="passthrough"
)
```

```python
x_trf1 = trf1.fit_transform(x)
```

```python
x_trf1_df = pd.DataFrame(x_trf1,columns=trf1.get_feature_names_out())
```

```python
x_trf2 = trf2.fit_transform(x_trf1)
```

```python
x_trf2_df =pd.DataFrame(x_trf2,columns=trf2.get_feature_names_out())
```

# Predefine Linear Regression

```python
import pandas as pd
import numpy as np 
from sklearn.datasets import load_diabetes
```

```python
x,y=load_diabetes(return_X_y=True,scaled=False)
```

```python
x.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30)
```

```python
x_train.shape
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
pred = lr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,root_mean_squared_error
```

```python
print("MAE : ", mean_absolute_error(pred, y_test))
print("MSE : ", mean_squared_error(pred,y_test))
print("RMSE : ", np.sqrt(mean_squared_error(pred,y_test)))
print("RMSE 1 : ", root_mean_squared_error(pred,y_test))
print("R2 Score : ", r2_score(pred,y_test))
```

# Custom Linear Regression

```python
class lllrrr():
    def __init__(self):
        coef_ = None
        intercept_ = None
        
    def fit(self,x_tr,y_tr):
        x_tr=np.insert(x_tr,0,1,axis=1)
        # print(x_tr.shape,type(x_tr))

        betas = np.linalg.inv(np.dot(x_tr.T,x_tr)).dot(x_tr.T).dot(y_tr)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]
        print(self.intercept_,self.coef_,sep='\n')
        
    
    def predict(self,x_tes):
        y_pr=np.dot(x_tes,self.coef_) + self.intercept_
        return y_pr
```

```python
lr1 = lllrrr()
lr1.fit(x_train,y_train)
```

```python
pred = lr1.predict(x_test)
```

```python
print("MAE : ", mean_absolute_error(pred, y_test))
print("MSE : ", mean_squared_error(pred,y_test))
print("RMSE : ", np.sqrt(mean_squared_error(pred,y_test)))
print("RMSE 1 : ", root_mean_squared_error(pred,y_test))
print("R2 Score : ", r2_score(pred,y_test))
```

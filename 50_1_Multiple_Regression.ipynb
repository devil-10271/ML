```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from sklearn.linear_model import LinearRegression 
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
import plotly.graph_objects as go
```

```python
x,y=make_regression(n_samples=1000, n_features=2, n_informative=2, n_targets=1, noise=50)
```

```python
df=pd.DataFrame({'X1':x[:,0],'X2':x[:,1],"Y":y})
```

```python
fig = px.scatter_3d(x=df.X1,y=df.X2,z=df.Y)
fig.update_layout(margin=dict(l=20,r=20,t=20,b=20))
fig.show()
```

```python
x_tr,x_te,y_tr,y_te=train_test_split(df.drop("Y",axis=1),df.Y,test_size=0.2)
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_tr,y_tr)
```

```python
pred = lr.predict(x_te)
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error,mean_absolute_error,mean_squared_error
```

```python
print("MAE : ", mean_absolute_error(pred, y_te))
print("MSE : ", mean_squared_error(pred,y_te))
print("RMSE : ", np.sqrt(mean_squared_error(pred,y_te)))
print("R2 Score : ", r2_score(pred,y_te))
r2 = r2_score(pred,y_te)
```

```python
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)
xGrid, yGrid = np.meshgrid(y, x)
final = np.vstack((xGrid.ravel().reshape(1,100),yGrid.ravel().reshape(1,100))).T

z_final = lr.predict(final).reshape(10,10)

z = z_final
```

```python
fig = px.scatter_3d(df, x='X1', y='X2', z='Y')

fig.add_trace(go.Surface(x = x, y = y, z =z ))
fig.update_layout(margin=dict(l=20,r=20,t=20,b=20))
fig.show()
```
